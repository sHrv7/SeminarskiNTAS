```python
class BinaryNode:

    def __init__ (self, value = None):
        """Create binary node."""
        self.value  = value
        self.left   = None
        self.right  = None
        self.height = 0
        
        
    def computeHeight (self):
        """Compute height of node in BST."""
        height = -1
        if self.left:
            height = max(height, self.left.height)
        if self.right:
            height = max(height, self.right.height)
            
        self.height = height + 1

    def dynamicHeight (self):
        """Compute height of node in BST."""
        height = -1
        if self.left:
            height = max(height, self.left.dynamicHeight())
        if self.right:
            height = max(height, self.right.dynamicHeight())
            
        return height + 1

    def dynamicHeightDifference (self):
        """Compute height difference of node's children in BST."""
        leftTarget = 0
        rightTarget = 0
        if self.left:
            leftTarget = 1 + self.left.dynamicHeight()
        if self.right:
            rightTarget = 1 + self.right.dynamicHeight()
            
        return leftTarget - rightTarget

    def heightDifference (self):
        """Compute height difference of node's children in BST."""
        leftTarget = 0
        rightTarget = 0
        if self.left:
            leftTarget = 1 + self.left.height
        if self.right:
             rightTarget = 1 + self.right.height
                   
        return leftTarget - rightTarget

    def assertAVLProperty (self):
        """Validate AVL property for BST node."""
        if abs(self.dynamicHeightDifference()) > 1:
            return False
        if self.left:
            if not self.left.assertAVLProperty():
                return False
        if self.right:
            if not self.right.assertAVLProperty():
                return False

        return True

    def rotateRight (self):
        """Perform right rotation around given node."""
        newRoot = self.left
        grandson = newRoot.right
        self.left = grandson
        newRoot.right = self

        self.computeHeight()
        
        return newRoot

    def rotateLeft (self):
        """Perform left rotation around given node."""
        newRoot = self.right
        grandson = newRoot.left
        self.right = grandson
        newRoot.left = self
    
        self.computeHeight()
        return newRoot

    def rotateLeftRight (self):
        """Perform left, then right rotation around given node."""
        child = self.left
        newRoot = child.right
        grand1  = newRoot.left
        grand2  = newRoot.right
        child.right = grand1
        self.left = grand2
    
        newRoot.left = child
        newRoot.right = self
    
        child.computeHeight()
        self.computeHeight()
        
        return newRoot

    def rotateRightLeft (self):
        """Perform right, then left rotation around given node."""
        child = self.right
        newRoot = child.left
        grand1  = newRoot.left
        grand2  = newRoot.right
        child.left = grand2
        self.right = grand1
    
        newRoot.left = self
        newRoot.right = child
    
        child.computeHeight()
        self.computeHeight()
      
        return newRoot
    
    def compareTo (self, value):
        """
        Returns 0 if equal, negative if smaller and positive if greater.
        Suitable for overriding.
        """
        if self.value == value:
            return 0
        if self.value < value:
            return -1
        return +1
        
    def add (self, val):
        """Adds a new node to BST with value and rebalance as needed."""
        newRoot = self

        # if val <= self.value        
        if self.compareTo(val) >= 0:
            self.left = self.addToSubTree (self.left, val)
            if self.heightDifference() == 2:
                #if val <= self.left.value:
                
                if self.left.compareTo(val) >= 0:
                    newRoot = self.rotateRight()
                else:
                    newRoot = self.rotateLeftRight()
        else:
            self.right = self.addToSubTree (self.right, val)
            if self.heightDifference() == -2:
                #if val > self.right.value:
                if self.right.compareTo(val) < 0:
                    newRoot = self.rotateLeft()
                else:
                    newRoot = self.rotateRightLeft()

        newRoot.computeHeight()
        return newRoot

    def newNode (self, val):
        """Return new Binary Node, amenable to subclassing."""
        return BinaryNode(val)

    def addToSubTree (self, parent, val):
        """Add val to parent subtree (if exists) and return root in case it has changed because of rotation."""
        if parent is None:
            return self.newNode(val)

        parent = parent.add(val)
        return parent
           
    def removeFromParent (self, parent, val):
        """Helper method for remove. Ensures proper behavior when removing node that 
        has children."""
        if parent:
            return parent.remove(val)
        return None

    def remove (self, val):
        """
         Remove val from BinaryTree. Works in conjunction with remove
         method in BinaryTree.
        """
        newRoot = self
        rc = self.compareTo(val)
        if rc == 0:
            if self.left is None:
                return self.right

            child = self.left
            while child.right:
                child = child.right
            
            childKey = child.value;
            self.left = self.removeFromParent(self.left, childKey)
            self.value = childKey;

            if self.heightDifference() == -2:
                if self.right.heightDifference() <= 0:
                    newRoot = self.rotateLeft()
                else:
                    newRoot = self.rotateRightLeft()
        elif rc > 0:
            self.left = self.removeFromParent(self.left, val)
            if self.heightDifference() == -2:
                if self.right.heightDifference() <= 0:
                    newRoot = self.rotateLeft()
                else:
                    newRoot = self.rotateRightLeft()
        else:
            self.right = self.removeFromParent(self.right, val)
            if self.heightDifference() == 2:
                if self.left.heightDifference() >= 0:
                    newRoot = self.rotateRight()
                else:
                    newRoot = self.rotateLeftRight()

        newRoot.computeHeight()
        return newRoot

    def __repr__ (self):
        """Useful debugging function to produce linear tree representation."""
        leftS = ''
        rightS = ''
        if self.left:
            leftS = str(self.left)
        if self.right:
            rightS = str(self.right)
        return "(L:" + leftS + " " + str(self.value) + " R:" + rightS + ")"

    def inorder (self):
        """In order traversal generator of tree rooted at given node."""
        if self.left:
            for n in self.left.inorder():
                yield n

        yield self.value

        if self.right:
            for n in self.right.inorder():
                yield n

class BinaryTree:

    def __init__ (self):
        """Create empty binary tree."""
        self.root = None

    def __repr__ (self):
        if self.root is None:
            return "avl:()"
        return "avl:" + str(self.root)
                
    def newNode (self, value):
        """Return new BinaryNode object. Suitable for overriding."""
        return BinaryNode(value)
    
    def add (self, value):
        """Insert value into proper location in Binary Tree."""
        if self.root is None:
            self.root = self.newNode(value)
        else:
            self.root = self.root.add(value)

    def __contains__ (self, target):
        """Check whether BST contains target value."""
        node = self.root
        while node:
            rc = node.compareTo(target)
            if rc > 0:
                node = node.left
            elif rc < 0:
                node = node.right
            else:
                return True
        return False

           
    def remove (self, val):
        """Remove value from tree."""
        if self.root:
            self.root = self.root.remove(val)

    def __iter__ (self):          # definira iteriranje kroz inorder ispis
        """In order traversal of elements in the tree."""
        if self.root:
            return self.root.inorder()
                        
    def assertAVLProperty (self):
        """Validate AVL property for BST Tree."""
        if self.root:
            return self.root.assertAVLProperty()
        else:
            return True
        
    
    def _compare_keys(self, person, key):
        """Uspoređuje osobu s ključem (prezime, ime)."""
        if isinstance(key, str):
            return (person.prezime > key) - (person.prezime < key)
        elif isinstance(key, tuple) and len(key) == 2:
            prezime_cmp = (person.prezime > key[0]) - (person.prezime < key[0])
            if prezime_cmp != 0:
                return prezime_cmp
            return (person.ime > key[1]) - (person.ime < key[1])
        else:
            raise ValueError("Invalid key format")
```

```python
class PersonAVLNode(BinaryNode):
    def compareTo(self, other):
        """
        Uspoređuje osobe po prezimenu, pa po imenu.
        """
        if not isinstance(other, Osoba):
            return super().compareTo(other)
        
        # Usporedba po prezimenu
        prezime_cmp = (self.value.prezime > other.prezime) - (self.value.prezime < other.prezime)
        if prezime_cmp != 0:
            return prezime_cmp
        
        # Ako prezimena jednaka, usporedba po imenu
        ime_cmp = (self.value.ime > other.ime) - (self.value.ime < other.ime)
        return ime_cmp

class PersonBinaryNode(BinaryNode):
    """Obični BST čvor bez rotacija za osobe."""
    
    def add(self, val):
        """Dodaje čvor bez balansiranja."""
        if self.compareTo(val) >= 0:
            if self.left is None:
                self.left = self.newNode(val)
            else:
                self.left = self.left.add(val)
        else:
            if self.right is None:
                self.right = self.newNode(val)
            else:
                self.right = self.right.add(val)
        
        self.computeHeight()
        return self
    
    def compareTo(self, other):
        """
        Uspoređuje osobe po prezimenu, pa po imenu.
        """
        if not isinstance(other, Osoba):
            return super().compareTo(other)
        
        # Usporedba po prezimenu
        prezime_cmp = (self.value.prezime > other.prezime) - (self.value.prezime < other.prezime)
        if prezime_cmp != 0:
            return prezime_cmp
        
        # Ako prezimena jednaka, usporedba po imenu
        ime_cmp = (self.value.ime > other.ime) - (self.value.ime < other.ime)
        return ime_cmp
    
    def newNode(self, val):
        """Vraća novi PersonBinaryNode."""
        return PersonBinaryNode(val)

class BinarySearchTree(BinaryTree):
    """Obično binarno stablo pretraživanja bez balansiranja za osobe."""
    
    def newNode(self, value):
        return PersonBinaryNode(value)
```

```python
class PersonAVLNode(BinaryNode):
    # Globalni brojač rotacija
    rotation_count = 0
    left_rotation_count = 0
    right_rotation_count = 0
        
    def rotateRight(self):
        """Perform right rotation around given node."""
        PersonAVLNode.rotation_count += 1
        PersonAVLNode.right_rotation_count += 1
        return super().rotateRight()

    def rotateLeft(self):
        """Perform left rotation around given node."""
        PersonAVLNode.rotation_count += 1
        PersonAVLNode.left_rotation_count += 1
        return super().rotateLeft()

    def rotateLeftRight(self):
        """Perform left, then right rotation around given node."""
        PersonAVLNode.rotation_count += 2
        PersonAVLNode.left_rotation_count += 1
        PersonAVLNode.right_rotation_count += 1
        return super().rotateLeftRight()

    def rotateRightLeft(self):
        """Perform right, then left rotation around given node."""
        PersonAVLNode.rotation_count += 2
        PersonAVLNode.right_rotation_count += 1
        PersonAVLNode.left_rotation_count += 1
        return super().rotateRightLeft()

class PersonBinaryNode(BinaryNode):
    """Obični BST čvor bez rotacija za osobe."""
    
    def add(self, val):
        """Dodaje čvor bez balansiranja."""
        if self.compareTo(val) >= 0:
            if self.left is None:
                self.left = self.newNode(val)
            else:
                self.left = self.left.add(val)
        else:
            if self.right is None:
                self.right = self.newNode(val)
            else:
                self.right = self.right.add(val)
        
        self.computeHeight()
        return self
    
    def newNode(self, val):
        """Vraća novi PersonBinaryNode."""
        return PersonBinaryNode(val)

class BinarySearchTree(BinaryTree):
    """Obično binarno stablo pretraživanja bez balansiranja za osobe."""
    
    def newNode(self, value):
        return PersonBinaryNode(value)
    
class AVLSearchTree(BinaryTree):
    """AVL balansirano stablo pretraživanja za osobe."""
    
    def newNode(self, value):
        return PersonAVLNode(value)
    
    def get_rotation_stats(self):
        """Vraća statistiku rotacija."""
        return {
            'total': PersonAVLNode.rotation_count,
            'left': PersonAVLNode.left_rotation_count,
            'right': PersonAVLNode.right_rotation_count
        }
    
    def reset_rotation_stats(self):
        """Resetira brojače rotacija."""
        PersonAVLNode.rotation_count = 0
        PersonAVLNode.left_rotation_count = 0
        PersonAVLNode.right_rotation_count = 0
```

### a) Učitavanje podataka

```python
import csv
from collections import namedtuple

# Definiramo strukturu za osobu
Osoba = namedtuple('Osoba', ['ime', 'prezime', 'spol', 'datum_rodenja', 'mjesto_stanovanja'])

def ucitaj_podatke(datoteka):
    """Učitava podatke iz CSV datoteke i vraća listu objekata Osoba."""
    osobe = []
    with open(datoteka, 'r', encoding='utf-8') as f:
        reader = csv.reader(f)
        for row in reader:
            if len(row) == 5:
                osoba = Osoba(*(value.strip() for value in row))
                osobe.append(osoba)
    return osobe


def format_person_name(person):
    """Vraća prikaz prezimena i imena s oznakom za prazno ime."""
    ime = person.ime.strip()
    if ime == "":
        ime = "<prazno>"
    return f"{person.prezime}, {ime}"

# Učitavanje podataka

datoteka = 'podatkovni_skup_Zirdum.csv'
osobe = ucitaj_podatke(datoteka)
print(f"Učitano {len(osobe)} osoba.")

for osoba in osobe[:5]:  # Prikaz prvih 5 osoba
    print(osoba)
```

```python
def get_min_max_from_tree(tree):
    """Pronalazi minimalnu i maksimalnu vrijednost u stablu (inorder traversal)."""
    if not tree.root:
        return None, None
    
    values = list(tree)
    return values[0], values[-1]

# Pronalaženje min/max za BST
bst_min, bst_max = get_min_max_from_tree(bst_tree)
print("=" * 60)
print("MINIMALNA I MAKSIMALNA VRIJEDNOST INDEKSA")
print("=" * 60)
print("\nObično BST:")
if bst_min:
    print(f"Minimalna vrijednost (prezime): {format_person_name(bst_min)}")
    print(f"  Podaci: Ime={bst_min.ime or '<prazno>'}, Prezime={bst_min.prezime}, Spol={bst_min.spol}, Datum={bst_min.datum_rodenja}, Mjesto={bst_min.mjesto_stanovanja}")
if bst_max:
    print(f"Maksimalna vrijednost (prezime): {format_person_name(bst_max)}")
    print(f"  Podaci: Ime={bst_max.ime or '<prazno>'}, Prezime={bst_max.prezime}, Spol={bst_max.spol}, Datum={bst_max.datum_rodenja}, Mjesto={bst_max.mjesto_stanovanja}")

# Pronalaženje min/max za AVL
avl_min, avl_max = get_min_max_from_tree(avl_tree)
print("\nAVL stablo:")
if avl_min:
    print(f"Minimalna vrijednost (prezime): {format_person_name(avl_min)}")
    print(f"  Podaci: Ime={avl_min.ime or '<prazno>'}, Prezime={avl_min.prezime}, Spol={avl_min.spol}, Datum={avl_min.datum_rodenja}, Mjesto={avl_min.mjesto_stanovanja}")
if avl_max:
    print(f"Maksimalna vrijednost (prezime): {format_person_name(avl_max)}")
    print(f"  Podaci: Ime={avl_max.ime or '<prazno>'}, Prezime={avl_max.prezime}, Spol={avl_max.spol}, Datum={avl_max.datum_rodenja}, Mjesto={avl_max.mjesto_stanovanja}")
```

### e) Ispis stabla po razinama (prvih 10 razina)

```python
from collections import deque

def print_tree_by_levels(tree, max_levels=10):
    """Ispisuje stablo po razinama (level-order traversal) do max_levels razina."""
    if not tree.root:
        print("Stablo je prazno.")
        return
    
    queue = deque([(tree.root, 0)])  # (čvor, razina)
    current_level = 0
    level_nodes = []
    
    while queue and current_level < max_levels:
        node, level = queue.popleft()
        
        if level > current_level:
            # Ispisuj prethodnu razinu
            print(f"Razina {current_level}: {level_nodes}")
            level_nodes = []
            current_level = level
        
        if current_level < max_levels:
            # Dodaj prezime i ime čvora
            level_nodes.append(f"{node.value.prezime},{node.value.ime}")
            
            if node.left:
                queue.append((node.left, level + 1))
            if node.right:
                queue.append((node.right, level + 1))
    
    if level_nodes:
        print(f"Razina {current_level}: {level_nodes}")

print("\n" + "=" * 60)
print("ISPIS STABLA PO RAZINAMA (PRVIH 10 RAZINA)")
print("=" * 60)

print("\nObično BST:")
print_tree_by_levels(bst_tree, max_levels=10)

print("\nAVL stablo:")
print_tree_by_levels(avl_tree_counted, max_levels=10)
```
